        # Program: Cat food Inventory
##### This program tracks, measures, and organize cat food stock so a cat shelter or store never runs out of food. 

```python
# Creating a list collection to hold cat food item weights in grams
food_bags = [430, 1300, 400, 300, 500]

# Changing the weight of the third bag (spot 2) because a cat ate some
print("Old weight at spot 2:", food_bags[2])
food_bags[2] = 200
print("New weight at spot 2:", food_bags[2])

# append() adds a new food bag and remove() deletesan empty food bag.
food_bags.append(1000)
print("Inventory after buying a bag:", food_bags)

food_bags.remove(200)
print("Inventory after removing empty a bag:", food_bags)

total_bags = len(food_bags)    # Counts the the food bags
smallest_bag = min(food_bags)  # Finds the smallest food bag
largest_bag = max(food_bags)   # Finds the largest food bag
total_weght = sum(food_bags)   # Adds all food weights

# Arranging the foods from the light to heavy
food_bags.sort()
print("Food stock sorted from low to high:", food_bags)

# Pointing directly to the last item to find the heaviest bag of food
heaviest_bag = food_bags[-1]
print("The single heaviest food bag is:", heaviest_bag)

# Looking at every single food bag in the inventory one by one 
print("---- Scanning Food Stock ----")
for weight in food_bags:
    print("Found cat food weighing:", weight, "grams")

# Splitting a text line of food flavors into lists of words
flavor_text = "Salmon,Chicken,Tuna,Beef,Mackerel"
flavor_list = flavor_text.split(",")

print("Raw text line of single flavors:", flavor_text)
print("Clean list of single flavors:", flavor_list)

# Going through the new flavor list 
for flavor in flavor_list:
    print("Available food flavor logged:", flavor)
```
